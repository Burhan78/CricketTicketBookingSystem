```python

# Ticket data

tickets = [
    [101, "VIP", "Available"],
    [102, "VIP", "Available"],
    [103, "General", "Available"],
    [104, "General", "Available"],
    [105, "Premium", "Available"],
    [106, "Premium", "Available"],
    [107, "General", "Available"],
    [108, "VIP", "Available"],
]

# Helper functions

def print_table_header():
    print(f"{'Seat Number':<14}{'Seat Name':<14}{'Status':<10}")
    print("-" * 38)


def print_ticket_row(ticket):
    print(f"{ticket[0]:<14}{ticket[1]:<14}{ticket[2]:<10}")


def find_seat(seat_number):
    """Linear search through the list.
    Returns the list index of the seat, or -1 if the seat does not exist."""
    for i in range(len(tickets)):
        if tickets[i][0] == seat_number:
            return i
    return -1


def count_by_status(status):
    """Count how many tickets currently have the given status."""
    count = 0
    for ticket in tickets:
        if ticket[2] == status:
            count += 1
    return count
```

```python
# Task 1: Display all tickets

def display_all_tickets():
    print("\n--- All Tickets ---")
    print_table_header()
    for ticket in tickets:
        print_ticket_row(ticket)
```

```python
# Task 2  Book one or more tickets

def book_tickets():
    print("\n--- Book a Ticket ---")
    available = count_by_status("Available")

    # nothing left to book
    if available == 0:
        print("Sorry! All tickets are sold out.")
        return

    # Show the customer which seats they can choose from
    display_available_tickets()
    print(f"\nTickets available: {available}")

    # No of tickets you want to buy?
    quantity_text = input("How many tickets do you want to buy? ").strip()
    if not quantity_text.isdigit():
        print("Invalid input! Please enter a whole number.")
        return

    quantity = int(quantity_text)
    if quantity < 1:
        print("Invalid input! You must buy at least 1 ticket.")
        return

    # cannot book more than what is available
    if quantity > available:
        print(f"Sorry! Only {available} ticket(s) are available. "
              f"You cannot book {quantity}.")
        return

    booked = 0
    while booked < quantity:
        if quantity > 1:
            print(f"\nTicket {booked + 1} of {quantity}")

        seat_text = input("Enter seat number (0 to stop booking): ").strip()

        # Input validation: must be a whole number
        if not seat_text.isdigit():
            print("Invalid input! Please enter a numeric seat number.")
            continue

        seat_number = int(seat_text)

        # Let the customer leave the transaction early
        if seat_number == 0:
            print("Booking stopped.")
            break

        index = find_seat(seat_number)

        if index == -1:
            # Seat number not in the list
            print("Invalid seat number! Please choose a seat from the list.")
        elif tickets[index][2] == "Sold":
            # Seat exists but is already taken
            print("Sorry! This ticket has already been sold.")
        else:
            # Seat exists and is available -> update the ORIGINAL list
            tickets[index][2] = "Sold"
            booked += 1
            print("Ticket Booked Successfully!")
            print(f"Seat Number: {tickets[index][0]}")
            print(f"Seat Name: {tickets[index][1]}")
            print(f"Status: {tickets[index][2]}")

    if quantity > 1:
        print(f"\nTransaction complete: {booked} of {quantity} ticket(s) booked.")
```

```python
# Task 3: Display available tickets

def display_available_tickets():
    print("\n--- Available Tickets ---")
    if count_by_status("Available") == 0:
        print("Sorry! No tickets are available.")
        return
    print_table_header()
    for ticket in tickets:
        if ticket[2] == "Available":
            print_ticket_row(ticket)
```

```python
# Task 4: Display sold tickets

def display_sold_tickets():
    print("\n--- Sold Tickets ---")
    if count_by_status("Sold") == 0:
        print("No tickets have been sold yet.")
        return
    print(f"{'Seat Number':<14}{'Seat Name':<14}")
    print("-" * 28)
    for ticket in tickets:
        if ticket[2] == "Sold":
            print(f"{ticket[0]:<14}{ticket[1]:<14}")
```

```python
# Task 5 Ticket Summary

def display_remaining_tickets():
    total = len(tickets)
    sold = count_by_status("Sold")
    available = count_by_status("Available")
    percent_sold = sold / total * 100

    print("\n--- Ticket Summary ---")
    print(f"Total Tickets: {total}")
    print(f"Sold Tickets: {sold}")
    print(f"Available Tickets: {available}")
    print(f"Percentage Sold: {percent_sold:.1f}%")
```

```python
# Menu

def main():
    running = True

    while running:
        print()
        print("=================================")
        print("      CRICKET TICKET SYSTEM")
        print("=================================")
        print("1. Display All Tickets")
        print("2. Book a Ticket")
        print("3. Display Available Tickets")
        print("4. Display Sold Tickets")
        print("5. Remaining Tickets")
        print("6. Exit")

        choice = input("Enter your choice: ")

        if choice == "1":
            display_all_tickets()
        elif choice == "2":
            book_tickets()
        elif choice == "3":
            display_available_tickets()
        elif choice == "4":
            display_sold_tickets()
        elif choice == "5":
            display_remaining_tickets()
        elif choice == "6":
            print("\nThank you for using the Cricket Ticket System. Goodbye!")
            running = False
        else:
            print("\nInvalid choice! Please enter a number between 1 and 6.")


main()
```
